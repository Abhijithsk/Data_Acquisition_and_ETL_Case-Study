#Task-5

```python
pop=pd.read_csv('/content/drive/MyDrive/DSAB13/Class_Activity/casestudylibraries/population.csv')

gdp=pd.read_excel('/content/drive/MyDrive/DSAB13/Class_Activity/casestudylibraries/gdp.xlsx')

net=pd.read_json('/content/drive/MyDrive/DSAB13/Class_Activity/casestudylibraries/internet_users.json')

lit=pd.read_xml('/content/drive/MyDrive/DSAB13/Class_Activity/casestudylibraries/literacy_rate.xml')

print('Number of rows :',pop.shape[0],'\n Number of columns :', pop.shape[1],
      '\n Column names:', pop.columns, '\n Data types:', pop.dtypes,
      '\n Missing values:', pop.isnull().sum(), '\n Duplicate records:',
      pop.duplicated().sum())
print('\n')
print('Number of rows :',gdp.shape[0],'\n Number of columns :', gdp.shape[1],
      '\n Column names:', gdp.columns, '\n Data types:', gdp.dtypes,
      '\n Missing values:', gdp.isnull().sum(), '\n Duplicate records:',
      gdp.duplicated().sum())
print('\n')

print('Number of rows :',net.shape[0],'\n Number of columns :', net.shape[1],
      '\n Column names:', net.columns, '\n Data types:', net.dtypes,
      '\n Missing values:', net.isnull().sum(), '\n Duplicate records:',
      net.duplicated().sum())
print('\n')

print('Number of rows :',lit.shape[0],'\n Number of columns :', lit.shape[1],
      '\n Column names:', lit.columns, '\n Data types:', lit.dtypes,
      '\n Missing values:', lit.isnull().sum(), '\n Duplicate records:',
      lit.duplicated().sum())


```

#Task-6,7,8,9

```python
gdp = gdp.rename(columns={"Country": "country", "GDP": "gdp"})
lit = lit.rename(columns={"name": "country"})
lit["literacy_rate"] = pd.to_numeric(lit["literacy_rate"])

df = pop.merge(gdp, on="country").merge(net, on="country").merge(lit, on="country")

df["internet_penetration"] = (df["internet_users"] / df["population" ] * 100).round(2)
df["gdp_per_capita"] = (df["gdp"] * 1000000 / df["population"]).round(2)
df
```

```python
print(pop.dtypes)
print(gdp.dtypes)
print(net.dtypes)
print(lit.dtypes)
```

#Task-10,11

```python
import sqlite3
```

```python
conn = sqlite3.connect("country.db")
df.to_sql("country_statistics", conn, if_exists="replace", index=False)
print(pd.read_sql("SELECT * FROM country_statistics LIMIT 5", conn))
```

#Task-12

```python
print(df.nlargest(5, "gdp")[["country", "gdp"]])
print('\n')

print(df.nlargest(5, "population")[["country", "population"]])
print('\n')

print(df["literacy_rate"].mean())
print('\n')

print(df[df["literacy_rate"] > 90]["country"].tolist())
print('\n')

print(df[df["internet_penetration"] > 70]["country"].tolist())
print('\n')

print(df.nlargest(5, "gdp_per_capita")[["country", "gdp_per_capita"]])
print('\n')

print(df["population"].sum())
print('\n')

print(df["internet_penetration"].mean())
print('\n')

print(df.loc[df["internet_users"].idxmax(), "country"])
print('\n')

print(df["literacy_rate"].corr(df["internet_penetration"]))
```

#Part F

```python

# 11
print(pd.read_sql("SELECT country, gdp FROM country_statistics ORDER BY gdp DESC LIMIT 5", conn))
print('\n')

# 12
print(pd.read_sql("SELECT ROUND(AVG(literacy_rate), 2) AS avg_literacy FROM country_statistics", conn))
print('\n')

# 13
print(pd.read_sql("SELECT country, literacy_rate FROM country_statistics WHERE literacy_rate > 90", conn))
print('\n')

# 14
print(pd.read_sql("SELECT country, internet_penetration FROM country_statistics ORDER BY internet_penetration DESC LIMIT 5", conn))
print('\n')

# 15
print(pd.read_sql("SELECT SUM(population) AS total_population FROM country_statistics", conn))
print('\n')

# 16
print(pd.read_sql("SELECT country, gdp_per_capita FROM country_statistics ORDER BY gdp_per_capita DESC LIMIT 1", conn))
print('\n')

# 17
pd.read_sql("SELECT * FROM country_statistics", conn).to_csv("country_statistics.csv", index=False)
print('\n')

# 18
print(pd.read_sql("SELECT country, gdp_per_capita FROM country_statistics ORDER BY gdp_per_capita DESC LIMIT 10", conn))
```
